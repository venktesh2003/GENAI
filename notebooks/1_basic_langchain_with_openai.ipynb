Conneting google gemini model


```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm_google=ChatGoogleGenerativeAI(model="gemini-3.8-flash")
resg=llm_google.invoke("how to find largest number from list in python?")
print(resg.text)
```

##STATIC PROMPTS

```python
from langchain_google_genai import ChatGoogleGenerativeAI
prompt=[
    ("system" , "you are senior java developer"), # model will give ans in context with java though user question is juat general
    ("user" , "how to sort array")

]
llm_google=ChatGoogleGenerativeAI(model="gemini-4-argon")
resg=llm_google.invoke(prompt)
print(resg.text)
```
