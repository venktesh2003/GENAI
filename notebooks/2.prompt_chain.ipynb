```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")


prompts=ChatPromptTemplate.from_messages([
  {"role":"system" , "content":"you are a translator translate into {language}"},
  {"role":"user" , "content":"{query}"}

])
finalprompt = prompts.format_messages(language="hindi",query="hey fat ass")
finalprompt
#res = llm.invoke(finalprompt)
#res
```

```python
#or we can format

res = llm.invoke(prompts.invoke({"language":"hindi" , "query":"i love cricket"}))
res.content
```

CHAIN WE DONT NEED ABOVE 2 IF WE CHAIN

```python
from langchain_core.output_parsers import StrOutputParser
out = StrOutputParser()

def transformcase(result):
    return result.upper()


chain = prompts | llm | out | transformcase     #we have chained out and function transform case
#first it excute prompts and than output of prompts become input to llm later we will invoke the chain by passing dict of our prompt

res = chain.invoke({"language":"marathi" , "query":"had your food "})
res
```
