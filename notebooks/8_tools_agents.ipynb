```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool

```

```python
@tool
def add_numbers(a:int , b:int):
    """
    it will return the sum of two numbers

    Args:
      a : number1
      b: number2

    """
    return a+b

@tool
def multiply_numbers(a:int , b:int):
    """
    it will return the sum of two numbers

    Args:
      a : number1
      b: number2

    """
    return a*b
```

```python
add_numbers.invoke({"a":13 , "b":5})
multiply_numbers.invoke({"a":13 , "b":5})  #calling tool with invoke
```

connect tool with LLM


```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
llm = ChatGroq(model="openai/gpt-oss-20b")
agent=create_agent(model=llm , tools=[add_numbers , multiply_numbers] , system_prompt="you are math teached and always use tool for calculation")
```

```python
res=agent.invoke({"messages":[{"role":"user" ,"content":"what is 2+3?"}]})
res
```

```python
#what is happening behind the scenes
for message in res['messages']:
    print(message)
    print("\n")
```
