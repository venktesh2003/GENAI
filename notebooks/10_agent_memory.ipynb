```python
from dotenv import load_dotenv

load_dotenv()
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
from langgraph.checkpoint.memory import InMemorySaver   # import the in memory saver  this will help to remember the prev convo
```

```python
llm=ChatGroq(model="openai/gpt-oss-20b")  #llm
search=GoogleSerperAPIWrapper()  
```

```python
#creating the agent using in memory saver
agent=create_agent(model=llm , 
tools=[search.run] , 
system_prompt="your are agent and you can search question on google" ,
checkpointer=InMemorySaver()  #this will make sure to remember our agent abot convo


)
```

```python
resp1=agent.invoke(
   { "messages":[{"role":"user" , "content":"who won the world cup women 2025??"}]} , 
   {"configurable":{"thread_id":1}}
)
```

```python
resp2=agent.invoke(
   { "messages":[{"role":"user" , "content":"what was score?"}]} , 
   {"configurable":{"thread_id":1}}
)
```
