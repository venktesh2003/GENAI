SPLITTING THE DOCUMENTS AND TEXT

```python
# we will use recursecharacter trext splitters
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = "Generative Artificial Intelligence (Gen AI) is a branch of Artificial Intelligence that can create new content such as text, images, audio, video, and code. Unlike traditional AI systems that mainly analyze or classify data, Gen AI generates new outputs based on patterns learned from large datasets."
```

```python
#create object of recursive character text splitter
splitter=RecursiveCharacterTextSplitter(chunk_size=50 , chunk_overlap=20)
```

```python
#splitting text
chunk=splitter.split_text(text)
chunk
```

```python
from langchain_community.document_loaders import PyPDFLoader,TextLoader
```

SPLITTING TXT DOCUMENTS INTO CHUNKS

```python
#loading and splitting text document
txt_loader = TextLoader("../apps/Data/sample.txt")
txt_chunk=txt_loader.load()
txt_chunk

#splitting the text chunk
txt_doc=splitter.split_documents(txt_chunk)
txt_doc
```

Loading and splitting PDF dat

```python
pdf_loader = PyPDFLoader("../apps/Data/Generative_AI_6_Page_Guide.pdf")
pdf_content=pdf_loader.load()
from langchain_text_splitters import RecursiveCharacterTextSplitter

#splitting the content into chunk
pdf_splitter = RecursiveCharacterTextSplitter(chunk_size=200 , chunk_overlap=50)
pdf_chunk=pdf_splitter.split_documents(pdf_content)
pdf_chunk
```
