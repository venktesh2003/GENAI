With Pydantic 

```python
from pydantic import BaseModel
```

```python
class User(BaseModel):
    #we can specify data type now
    name :str
    age :int
```

```python

user=User(name="venktesh" , age="my name is venktesh")  # this will give error as age data type is integer
```

```python
user=User(name="venktesh" , age=23)  # this is the strict check of pydantic
x = user.name + str(user.age)
x
```

FIELD VALIDATIONS

```python
from pydantic import Field
class Product(BaseModel):
    title:str=Field(max_length=10)
    des:str= Field(min_length=20)
    price:int = Field(gt=0 , le=100)
```

```python
pdct=Product(title="this is my" , des="sdjffdifdjfndifjndidnfdijvndivndvidnidfnddnvijndvdnd" ,price=40)
```

How to connect class to class

```python
class Order(BaseModel):
    id : str=Field(min_length=5 , max_length=10)
    product:Product   #this is the class which we created above now these two are linked
    
```

```python
order1=Order(id="32323" , product=Product(title="i17" , des="this awesome phone and its great",price=12))
order1
```

```python
order2=Order(id="3454545" , product=pdct)
order2
```

passing List type of class

```python
from typing import List
class Cart(BaseModel):
   id : str
   items : List[Product]
```

```python
crt1 = Cart(id="2323232" , items=[pdct , Product(title="i17" , des="this awesome phone and its great",price=12)])
```

```python
crt1
```

how to create common validator using Annotated


```python
from typing import Annotated 

commonvalidator = Annotated[List[Product] , Field(default=[])]
```

```python
from typing import List
class Cart(BaseModel):
   id : str
   items :commonvalidator #List[Product]  # this is same and you created common validator

crt1 = Cart(id="2323232" , items=[pdct , Product(title="i17" , des="this awesome phone and its great",price=12)])
crt1
```
