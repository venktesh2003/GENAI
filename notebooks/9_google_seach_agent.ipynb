```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm=ChatGroq(model="openai/gpt-oss-20b")  #llm
search=GoogleSerperAPIWrapper()           #serprer API object
```

```python
#How to search using serper API 
search.run("who is the founder of macro vision academy?")
```

```python
# create agent now 
agent=create_agent(model=llm , 
tools=[search.run] , 
system_prompt="your are agent and you can search question on google"


)
```

```python
question = "who is founder of all is well hospital burhanpur"
```

```python
response=agent.invoke({"messages":[{"role":"user" , "content":question}]})
```

```python
for res in response["messages"]:
    print(res ,end="\n")
```

```python
print(response["messages"][-1].content)
```
