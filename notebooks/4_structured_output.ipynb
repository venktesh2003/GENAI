```python
from dotenv import load_dotenv


load_dotenv()
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")

```

```python
from pydantic import BaseModel, Field
class Movies(BaseModel):
    name:str=Field(Description="name of movie")
    year:str=Field(description="year of release")

movie_llm = llm.with_structured_output(Movies)
res=movie_llm.invoke("give me one sci-fi movie")
res
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from pydantic import BaseModel ,Field 
from typing import List

class movie(BaseModel):
    name:str=Field(description="name of movie")
    actor:str=Field(description="name of actor")
    year:int=Field(description="year of release")

class Movies(BaseModel):
    m:List[movie]

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
moviellm = llm.with_structured_output(Movies)
res=moviellm.invoke("give me three hindi movies")
res
```
