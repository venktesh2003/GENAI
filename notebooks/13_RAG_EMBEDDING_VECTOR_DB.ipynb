VECTOR EMBEEDING or Vector generation

```python
from langchain_openai import OpenAIEmbeddings
from langchain_google_genai import GoogleGenerativeAIEmbeddings
#you can use OpenAI embedding or google embeeding 
```

```python
text = ["Deep learning is the neural network with multiple layers" , "Machine learning is the subset of AI" , "python is the programming language" , "India won world cup in 2022" , "AI model can understand the human languahge"]
```

```python
#create object of embedding
embd = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview")
vector=embd.embed_documents(text)
```

```python
vector
len(vector[0])
```

STORE VECTOR EMBEDDINGS IN VECTOR DB

```python
from langchain_chroma import Chroma
```

```python
vectore_store = Chroma.from_texts(
    texts=text ,   #what you need to embeed
    embedding=embd,   #which embedding model you wantt to use
    persist_directory="./vector_db"   # DB path where you want to store embeddings
)
```

Querying the data 

```python
query = "Machine learning"
result=vectore_store.similarity_search(query , k=2)
```
